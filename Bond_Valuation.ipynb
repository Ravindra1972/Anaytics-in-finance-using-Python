```python
def bond_price(face_value, coupon_rate, market_rate, years, payments_per_year=1):
    """
    Calculate the present value (price) of a bond.

    Parameters:
    - face_value: The bond's face/par value (e.g., 1000)
    - coupon_rate: Annual coupon rate (e.g., 0.05 for 5%)
    - market_rate: Market interest rate / YTM (e.g., 0.04 for 4%)
    - years: Time to maturity in years
    - payments_per_year: Number of coupon payments per year (default: 1 for annual)

    Returns:
    - price: Present value of the bond
    """
    periods = years * payments_per_year
    coupon_payment = face_value * coupon_rate / payments_per_year
    discount_rate = market_rate / payments_per_year

    # Present value of coupons
    pv_coupons = sum([
        coupon_payment / (1 + discount_rate) ** t
        for t in range(1, int(periods) + 1)
    ])

    # Present value of face value
    pv_face = face_value / (1 + discount_rate) ** periods

    price = pv_coupons + pv_face
    return price


# Example usage
if __name__ == "__main__":
    face_value = 1000
    coupon_rate = 0.05     # 5%
    market_rate = 0.04     # 4%
    years = 10
    payments_per_year = 2  # Semi-annual payments

    price = bond_price(face_value, coupon_rate, market_rate, years, payments_per_year)
    print(f"Bond Price: ${price:.2f}")
```
