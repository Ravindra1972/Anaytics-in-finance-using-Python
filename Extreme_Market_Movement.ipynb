```python
import yfinance as yf
```

```python
# Fetch historical data for the S&P 500
sp500 = yf.download('^GSPC', start='2023-01-01', end='2025-11-15')
```

```python
# Calculate daily returns
sp500['returns'] = sp500['Close'].pct_change()
```

```python
# Identify days with extreme price movements
extreme_days = sp500[abs(sp500['returns']) >
sp500['returns'].quantile(0.95)]
```

```python
print(f"Extreme market movements observed on {len(extreme_days)} days")
```
