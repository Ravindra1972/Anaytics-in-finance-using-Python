```python
import numpy as np

def call_payoff (S, K, premium =0):
    #""" Long call payoff ."""
    return np.maximum(S - K, 0) - premium

def put_payoff (S, K, premium =0):
    # """ Long put payoff ."""
    return np.maximum(K - S, 0) - premium

def forward_payoff (S, K):
    # """ Long forward payoff ."""
    return S - K

# Price range
S = np.linspace(50 , 150 , 200)
K = 100
```

```python
# === 1. Basic Payoffs ===
print (" === OPTION PAYOFF ANALYSIS ===\n")
# Call at different strikes
print ("--- Call Option at Expiry ---")
for spot in [80 , 90, 100 , 110 , 120]:
    payoff = max ( spot - K, 0)
    profit = payoff - 5 # premium = 5
    status = " ITM" if spot > K else \
    ("ATM" if spot == K else " OTM")
    print (f" S_T ={ spot }: Payoff =${ payoff :>3}, "
    f" Profit =${ profit : >4} [{ status }]")
```

```python
# === 2. Put- Call Parity Check ===
print ("\n--- Put- Call Parity Verification ---")
S0 = 100 # Current stock price
K = 100 # Strike
r = 0.05 # Risk - free rate
T = 0.25 # 3 months
C = 6.89 # Call premium ( hypothetical )
# Solve for put from parity
P_parity = C - S0 + K * np.exp (-r * T)
print (f" Stock = ${S0}")
print (f" Strike = ${K}")
print (f" Rate = {r:.1%} , T = {T :.2f} years ")
print (f" Call premium = ${C :.2f}")
print (f" Put premium ( parity ) = ${ P_parity :.2f}")
print (f" C - P = ${C - P_parity :.2f}")
print (f" S - K*exp(-rT) = "
f"${S0 - K*np.exp(-r*T):.2f}")
print (f" Match : {np.isclose(C - P_parity , S0 - K*np.exp(-r*T))}")
```

```python
# === 3. Combination Strategies ===
print ("\n--- Straddle ( Long Call + Long Put) ---")
straddle_cost = C + P_parity
print (f" Total premium : ${ straddle_cost :.2f}")
print (f" Breakeven (up): ${K + straddle_cost :.2f}")
print (f" Breakeven (dn): ${K - straddle_cost :.2f}")
print (f" Max loss : ${ straddle_cost :.2f} "
f"(if S_T = K = {K})")
print (f" Strategy : Bet on volatility , "
f"not direction ")
```

```python
# === 4. Forward Pricing ===
print ("\n--- Forward Pricing (No- Arbitrage ) ---")
S0_examples = [100 , 100 , 100]
r_examples = [0.05 , 0.05 , 0.10]
T_examples = [0.25 , 1.0 , 1.0]
for s, rate , t in zip( S0_examples , r_examples ,
    T_examples ):
    F = s * np.exp( rate * t)
    print (f" S0=${s}, r={ rate :.0%} , T={t}yr "
    f"-> Forward = ${F:.4f}")
```

```python
# === 5. Leverage Calculation ===
print ("\n--- Futures Leverage Example ---")
nifty = 25000
lot_size = 25
notional = nifty * lot_size
margin = 100000
leverage = notional / margin
print (f" Nifty : { nifty }")
print (f" Notional : Rs { notional :,}")
print (f" Margin : Rs { margin :,}")
print (f" Leverage : { leverage :.1f}x")
for move_pct in [1, 2, 5, -2, -5]:
    pnl = move_pct /100 * notional
    pnl_pct = pnl / margin * 100
    print (f" Nifty { move_pct :+d}%: "
    f"P&L = Rs { pnl :+,.0f} "
    f"({ pnl_pct :+.1f}% on margin )")
```
