```python
%%writefile app.py
#magic command,   writefile creates a new file
from sklearn.datasets import load_iris
from sklearn.neighbors import KNeighborsClassifier
import streamlit as st
import numpy as np
st.title('IRIS FLOWER CLASSIFICATION')
var = load_iris()
x = var.data  #inputs
y = var.target #outputs
model = KNeighborsClassifier(n_neighbors=13)
model.fit(x,y)
xmin = np.min(x, axis = 0) #min values of inputs to take all the features
xmax = np.max(x, axis =0)  #max values of input to take all the features

sepal_length = st.slider('Sepal Length',float(xmin[0]),float(xmax[0]))  # to set the proper range we use float(xmin) and float(xmax)
sepal_width = st.slider('Sepal width',float(xmin[1]),float(xmax[1]))
petal_length = st.slider('Petal Length',float(xmin[2]),float(xmax[2]))
petal_width = st.slider('Petal width',float(xmin[3]),float(xmax[3]))

y_pred = model.predict([[sepal_length,sepal_width,petal_length,petal_width]])
op = ['Setosa', 'Versicolor','Virginica']
st.title(op[y_pred[0]])  #we write 0 because the value will  be in 1 dimension
```
