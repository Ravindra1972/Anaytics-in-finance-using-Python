```python
# --- Market Cap calculator ---
def market_cap_category (price , shares_out ):
    """ Classify a stock by market cap."""
    cap = price * shares_out
    cap_bn = cap / 1e9
    if cap_bn > 10:
        category = " Large -Cap "
    elif cap_bn > 2:
        category = "Mid-Cap"
    elif cap_bn > 0.3:
        category = " Small -Cap "
    else :
        category = " Micro -Cap "
    return cap, category
```
