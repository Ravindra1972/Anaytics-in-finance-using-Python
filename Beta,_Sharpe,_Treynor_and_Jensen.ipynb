```python
import yfinance as yf
import numpy as np
import pandas as pd
import statsmodels.api as sm
```

```python
# Step 1: Download Data
stocks = ['SBIN.NS', 'TCS.NS', 'INFY.NS', 'BHEL.NS', 'CIPLA.NS', 'ITC.NS']  # portfolio/stock
benchmark = '^NSEI'  # S&P500 as market index

start_date = "2023-01-01"
end_date   = "2025-08-15"

df = yf.download(stocks + [benchmark], start=start_date, end=end_date)["Close"]
```

```python
# Step 2: Calculate Returns
returns = df.pct_change().dropna()
stock_ret = returns[stocks[0]]
market_ret = returns[benchmark]
```

```python
# Step 3: Set Risk-free Rate
rf = 0.04 / 252   # 6% annualized risk-free rate, daily
# Excess returns
excess_stock = stock_ret - rf
excess_market = market_ret - rf
```

```python
#Step 4: Sharpe Ratio
sharpe_ratio = (excess_stock.mean() / excess_stock.std()) * np.sqrt(252)
```

```python
#Step 5: Treynor Ratio
# (Need Beta first via regression)
# ------------------------------
X = sm.add_constant(market_ret)
model = sm.OLS(stock_ret, X).fit()
beta = model.params[1]

treynor_ratio = (stock_ret.mean() - rf) * 252 / beta   # annualized
```

```python
# Step 6: Jensen's Alpha

alpha = model.params[0] * 252   # annualized
```

```python
# Results
print(f"Sharpe Ratio: {sharpe_ratio:.4f}")
print(f"Treynor Ratio: {treynor_ratio:.4f}")
print(f"Jensen's Alpha: {alpha:.4f}")
print(f"Beta: {beta:.4f}")
```
